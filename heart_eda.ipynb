```python
hot_one_heart=pd.get_dummies(heart,columns=["ChestPainType","RestingECG","ExerciseAngina","ST_Slope","Sex"],drop_first=True)
hot_one_heart.astype('int')
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score,f1_score,confusion_matrix,classification_report

standard=StandardScaler()
columns=['Age', 'RestingBP', 'Cholesterol', 'FastingBS', 'MaxHR', 'Oldpeak',
       'ChestPainType_ATA', 'ChestPainType_NAP',
       'ChestPainType_TA', 'RestingECG_Normal', 'RestingECG_ST',
       'ExerciseAngina_Y', 'ST_Slope_Flat', 'ST_Slope_Up', 'Sex_M']
hot_one_heart[columns]=standard.fit_transform(hot_one_heart[columns])
```

```python
hot_one_heart
```

```python
hot_one_heart.columns
```

```python
x=hot_one_heart.drop(["HeartDisease"],axis=1)
y=hot_one_heart["HeartDisease"]
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
classification_models={
    "SVM_Model": SVC(kernel="rbf"),
    "Decision_tree Model":DecisionTreeClassifier(),
    "KNN_Model" :KNeighborsClassifier(),
    "Logistic_regession":LogisticRegression(),
    "Naive_Bayes_Model":GaussianNB()
}
```

```python
result=[]
```

```python
for name,model in classification_models.items():
    model.fit(x_train,y_train)
    y_predict= model.predict(x_test)

    accuracy=accuracy_score(y_test,y_predict)
    f1=f1_score(y_test,y_predict)
    confusionmetrix=confusion_matrix(y_test,y_predict)
    classification_reports=classification_report(y_test,y_predict)
    result.append({
        "Model":name,
        "Accuracy_score":round(accuracy,4),
        "F1_score":round(f1,4),
        # "Classification_report":classification_reports,
        # "Confusion_metrix":confusionmetrix
    })
    
```

```python
for i in result:
    print(i)
```

```python
import joblib
joblib.dump(classification_models["Logistic_regession"],"logistic_heart_model.pkl")
joblib.dump(standard,"scaler.pkl")
# joblib.dump(x.columns.to_list,"x_columns.pkl")
joblib.dump(x.columns.tolist(), "x_columns.pkl")
```
